```python
## we will have to handle all the null data points before we can dive into EDA and modeling
```

```python
Dataframe['Director'] = Dataframe['Director'].fillna('No Direction')
Dataframe['Cast'] = Dataframe['Cast'].fillna('No Cast')
Dataframe['Country'] = Dataframe['Country'].fillna('Country Unavailable')

# For dropna, we can use inplace=True safely on the main DataFrame, 
# or use assignment: Dataframe = Dataframe.dropna(subset=['Release_Date', 'Rating'])
Dataframe.dropna(subset=['Release_Date', 'Rating'], inplace=True)
```

```python
Dataframe.shape
```

```python
Dataframe.isnull().sum()
```

```python
## NOW WE WILL CREATE TWO COLUMNS Release_Datatime,Release_Year AND WE WILL GIVE DATETIME DATATYPE TO Release_Datetim
```

```python
Dataframe['Release_Datetime'] = pd.to_datetime(Dataframe['Release_Date'], format='mixed')
```

```python
Dataframe['Release_year']=Dataframe['Release_Datetime'].dt.year
```

```python
Dataframe.head()
```

```python
Dataframe.dtypes
```

```python
##NOW WE WILL CREATE TWO SEPARATE DATAFRAMES netflix_movies_df,netflix_shows_df AND PERFORM SOME OF THE DATA PREPARATION OPERATIONS TO IMPLEMENT THE DATAFRAMES
```

```python
netflix_movies_df=Dataframe[Dataframe['Category']=='Movie'].copy()
```

```python
netflix_movies_df.head(2)
```

```python
netflix_shows_df=Dataframe[Dataframe['Category']=='TV Show'].copy()
netflix_shows_df.head()
```

```python
netflix_shows_df.rename(columns={'Duration':'Seasons'},inplace=True)
netflix_shows_df.replace({'Seasons':{'1 Season':'1 Seasons'}},inplace=True)
netflix_shows_df.Seasons=netflix_shows_df.Seasons.str.replace('Seasons','').astype(int)
```

## TOTAL COUNT OF MOVIES AND TV SHOWS

```python
sns.set_theme(style='darkgrid')
```

```python
sns.countplot(x=Dataframe['Category'])
plt.title('Count of the Movies and TV Shows')
plt.xlabel('Category(Movies/TV Shows)')
plt.ylabel('Total Count')
```

```python
plt.figure(figsize=(12,6))
sns.countplot(data=Dataframe,x='Release_year',hue='Category')
plt.title('Year wise count of movies and tv shows');
```

```python
# 1. Clean and prepare data as a DataFrame instead of a standalone Series
filtered_countries = Dataframe.set_index('Category').Country.str.split(', ', expand=True).stack().reset_index(level=1, drop=True).reset_index(name='Country')

# 2. Filter out the missing/unavailable values
filtered_countries = filtered_countries[filtered_countries['Country'] != 'Country Unavailable']

# 3. Plot using data= and specifying x or y by column name
plt.figure(figsize=(12, 6))
sns.countplot(
    y='Country', 
    data=filtered_countries, 
    order=filtered_countries['Country'].value_counts().index[:20]
)
plt.title('Top 20 Countries on Netflix')
plt.xlabel('Count')
plt.ylabel('Country')
plt.show()
```

```python
# 1. Convert the series into a clean DataFrame with a named column
filtered_genre = Dataframe.set_index('Category').Type.str.split(', ', expand=True).stack().reset_index(level=1, drop=True).reset_index(name='Type')

# 2. Plot using the DataFrame approach
plt.figure(figsize=(12, 6))
sns.countplot(
    y='Type', 
    data=filtered_genre, 
    order=filtered_genre['Type'].value_counts().index[:20]
)
plt.title('Top 20 types on Netflix')
plt.xlabel('Count')
plt.ylabel('Types')
plt.show()
```

```python
order=['G','TV-Y','TV-G','PG','TV-Y7','TV-Y7-FV','TV-PG','PG-13','TV-14','R','NC-17','TV-MA']
plt.figure(figsize=(12,6))
sns.countplot(x=Dataframe.Rating,hue=Dataframe.Category,order=order);
plt.title('Rating for Movies and TV Shows')
plt.xlabel('Rating')
plt.ylabel('Total Count')
plt.show()
```

## NETFLIX MOVIES AND TV SHOWS DURATION

```python
fig,ax=plt.subplots(1,2,figsize=(19,5))
g1=sns.histplot(x=netflix_movies_df.Duration,color='Skyblue',ax=ax[0])
g1.set_xticks(np.arange(0, 331, 30))
g1.set_title('Duration Distribution for Netflix Movies')
g1.set_ylabel('Count of Netflix Movies')
g1.set_xlabel('Duration(minutes)')
g2=sns.countplot(x=netflix_shows_df.Seasons,color='Skyblue',ax=ax[1])
g2.set_title('Netflix TV Shows Seasons')
g2.set_ylabel('Count of netflix TV Shows')
g2.set_xlabel('Season(s)')
fig.show()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=Dataframe, x='Release_year', hue='Category', palette='Set2')
plt.title('Trend of Content Releases Over the Years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.xlim(max(0, len(Dataframe['Release_year'].unique()) - 15), None) # Show only the last 15 years for clarity
plt.show()
```

```python
plt.figure(figsize=(12, 6))
# Filter out 'Country Unavailable' for a cleaner geographical analysis
countries = Dataframe[Dataframe['Country'] != 'Country Unavailable']
top_10_countries = countries['Country'].value_counts().head(10)

sns.barplot(x=top_10_countries.values, y=top_10_countries.index, palette='viridis')
plt.title('Top 10 Content-Producing Countries on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=netflix_shows_df, x='Seasons', palette='magma')
plt.title('Distribution of TV Show Seasons')
plt.xlabel('Number of Seasons')
plt.ylabel('Count of TV Shows')
plt.show()
```
